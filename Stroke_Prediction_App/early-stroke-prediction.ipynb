```python
from sklearn.ensemble import StackingClassifier
from sklearn.linear_model import LogisticRegression
```

```python
base_models = [

    ('svm', SVC(
        probability=True,
        random_state=42
    )),

    ('rf', RandomForestClassifier(
        n_estimators=100,
        random_state=42
    )),

    ('xgb', XGBClassifier(
        n_estimators=100,
        learning_rate=0.1,
        max_depth=6,
        random_state=42
    )),

    ('lgbm', LGBMClassifier(
        n_estimators=100,
        learning_rate=0.1,
        random_state=42
    )),

    ('cat', CatBoostClassifier(
        iterations=100,
        learning_rate=0.1,
        depth=6,
        random_state=42,
        verbose=0
    ))

]
```

```python
meta_model = LogisticRegression()
```

```python
stack_model = StackingClassifier(

    estimators=base_models,

    final_estimator=meta_model,

    cv=5,

    n_jobs=-1

)
```

```python
stack_model.fit(X_train_smote, y_train_smote)
```

```python
y_pred_stack = stack_model.predict(X_test_scaled)

y_prob_stack = stack_model.predict_proba(X_test_scaled)[:,1]
```

```python
print("Accuracy:", accuracy_score(y_test, y_pred_stack))

print("Precision:", precision_score(y_test, y_pred_stack))

print("Recall:", recall_score(y_test, y_pred_stack))

print("F1 Score:", f1_score(y_test, y_pred_stack))

print("ROC-AUC:", roc_auc_score(y_test, y_prob_stack))
```
